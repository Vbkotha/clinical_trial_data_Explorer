# Trial Explorer: launch from a notebook

Run the cells in order with an **R kernel**. The app runs in a separate R process, so this notebook stays usable while the app is open.

Set `APP_DIR` below if you cloned the repository somewhere other than the parent of this folder.

```r
APP_DIR <- normalizePath("..")   # folder containing app.R
setwd(APP_DIR)
```

## 1. Install packages (first time only)

```r
source("install.R")
```

## 2. Run the unit tests (optional)

```r
testthat::test_file("tests/test_core.R")
```

## 3. Start the app

`role` is `"blinded"`, `"unblinded"` or `"viewer"` (completed studies). Leave it `NULL` for the most restrictive role for the study.

Studies in `config/studies.yml`: `STUDY-001` (ongoing, blinded views) and `STUDY-002` (completed, with final TFLs).

```r
source("launch.R")
run_trial_explorer(study = "STUDY-001", role = "blinded", port = 3838, background = TRUE)
```

Open the URL printed above. In Jupyter that is usually `<your session URL>/proxy/3838/`.

## 4. Stop the app

```r
stop_trial_explorer(port = 3838)
```
